```python
#task 1:Pandas Series Basics
#1.import pandas as pd
import pandas as pd

#2.Create a Pandas series from the list 
#3.Marks = [78, 85, 90, 66, 72]
Marks = [78, 85, 90, 66, 72]
series_values = pd.Series(Marks)

#4.print:
#series values 
#index
#Data type 
print("Series values ")
print(series_values)
print(f"series index:{series_values.index}")
print(f"Data type:{series_values.dtype}")

#5.Access:
#First element
#Last two elements
print(f"first element:{series_values[0]}")
print("last tow elements")
print(series_values.tail(2))
```

```python
#task 2:Mathematical operations on Series
#using the marks series:
#1.Add 5 grace marks to all students 
print("After Adding 5 grace marks to all student:")
print(series_values + 5)
#2.Subtract 2 marks from all values
print()
print("After Subtracting by 2 from all values:")
print(series_values - 2)
#3.Multiply all marks by 1.05
print()
print("After Multiplying by 1.05:")
print(series_values * 1.05)
#4.Divide all marks by 2
print()
print("After Divideing by 2:")
print(series_values/2)
#print result for each operation 
```

```python
import numpy as np
#Task 3:python functionalities on series 
#using the same series:
#1.find:
#maximum marks 
print("Maximum marks:",max(series_values))
#minimum marks 
print("minimum marks:",min(series_values))
#Sum of marks 
print("sum of marks:",sum(series_values))
#mean marks 
print("mean of marks:",np.mean(series_values))

#2.apply a lambda function to check whether each student has passed(>=70)
print()
print("student must get the 70 or more then 70 marks to pass")
print(Marks)
passed = lambda a: a >= 70
print(passed(series_values))

#3.Count how many students passed
print("No.of student passed:",np.count_nonzero(series_values >= 70))
```

```python

#task 4: Create a DataFrame
#Create a DataFrame students:
students = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}

#1.Convert it into a DataFrame
data = pd.DataFrame(students)
print("converted to dataFrame:")
print(data)

#2.print first 3 rows
print()
print("first 3 rows of the dataFrame:")
print(data.head(3))

#3.print last 2 rows
print()
print("last 2 rows of the dataFrame:")
print(data.tail(2))

#4.Print DataFrame shape and column names 
print()
print("dataFrame shape :",data.shape)
print("dataFrame coloumns:",data.columns)
```

```python
#Task 5:important DataFrame Functions
#Using the students DataFrame:
#1.Use and print output of:

#.info()
print("information of dataFrame:")
print(data.info())

#.describe()
print()
print("describeing about dataFrame:")
print(data.describe())

#.head()
print()
print("head of the DataFrame i am printing the first 2 rows:")
print(data.head(2))

#.tail()
print()
print("tail of the dataFrame i am printing the last 2 rows:")
print(data.tail(2))

#2.Sort students by Marks in descending order
df = data.sort_values(by='Marks',ascending=False)
print(df)

#3.Reset index after sorting 
print()
df = df.reset_index(drop=True)
print(df)
```

```python
#Task 6:Filtering & Conditional Selection 
#Perform the following filters
#1.Student who scored more than 75 marks 
print("marks more then 75")
print(data[data['Marks'] > 75])

#2.students belonging to subject math
print()
print("students belongs to maths")
print(data[data['Subject']== 'Math'])

#3.student who scored more than average marks 
print()
avg_marks =np.mean(data['Marks'])
print(f"average marks:{avg_marks}")
print("student who are get more then average marks")
print(data[data['Marks'] > avg_marks])

#4.Students who faild (marks < 70)
print()
print("students who are faild")
print(data[data['Marks'] < 70])
```

```python
#Task 7: Grouping & Basic Analysis
#1.find average marks per subject using groupby()
print("average marks per subject")
print(data.groupby('Subject')['Marks'].mean())

#2.Count number of students per subject
print()
print("number of students per subject")
print(data.groupby('Subject')['Name'].count())

#3.find maximum marks per subject.
print()
print("maximum marks per subject")
print(data.groupby('Subject')['Marks'].max())
```

```python
#Task 8: Pandas Plotting (Simple Graphs)
#Using Pandas built-in plotting:

#1.Plot a bar graph of student names vs marks.
bar =data.plot(x='Name', y='Marks', kind='bar')
print("bar graph")
print(bar)
```

```python
#2.plot a line graph of marks
line = data['Marks'].plot(kind='line')
print("line graph")
print(line)
```

```python
#3.plot a histogram of marks
hist =data['Marks'].plot(kind='hist')
print("histogram")
print(hist)

#Note: Use only:
#- DataFrame.plot()
#- Series.plot()
#(No matplotlib customization required.)
```
