```python
#Task->4 Generate Summary Report from file 
#Using only file reading operations:

#1->Read all sales values from sales_data.txt
#2->convert them into integers(i am using here map,list,split functions for the convertion string to int and creating a list)
with open('sales_data.txt', 'r') as f:
    content = f.read()
    string_to_numbers = list(map(int, content.split()))
    print(string_to_numbers)

    #3->Calculate and print:Total_sales, Highest_sale, Lowest_Sale, Average_Sale
    Total_sales =0
    for i in string_to_numbers:
        Total_sales += i 
    print('Total_sales:',  Total_sales)
    print('max value is:', max(string_to_numbers))
    print('min value is:', min(string_to_numbers))
    print('average Sale:', Total_sales/len(string_to_numbers))
```
