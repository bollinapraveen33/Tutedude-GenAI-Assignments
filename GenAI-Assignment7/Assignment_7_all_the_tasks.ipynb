```python
#Task-1 Basic Class & Object Creation 
#create a class Product with:
#Attributes:name, price, category

class Product:

    def __init__(self, name, price, category):
        self.name = name 
        self.price = price 
        self.category = category

#Method: get_info()--> prints product details 
    def get_info(self):
        print(f"name of the product: {self.name} with price: {self.price} and category is: {self.category}")

#Extra(optional):Add a method apply_discount(percent)that returns the discount price
    def apply_discount(self, percent):
        discounted_price = self.price - (self.price * percent)/100
        return discounted_price


#create two object and call get_info()
product_1 = Product("laptop", 75000, "electronics")
product_2 = Product("Keyboard", 2000, "electronics")
product_1.get_info()
product_2.get_info()
print("After applying the discount price is:", product_1.apply_discount(10))
print("After applying the discount price is:", product_2.apply_discount(10))

        
```

```python
#Task-2 Constructor & Encapsulation 
#Modify the Product class:
#Make price a private attribute(__price)

class Product:

    def __init__(self, name, price, category):
        self.name = name 
        self.__price = price 
        self.category = category

    #Create getter & setter methods:
    #getter method --> get_price()
    def get_price(self):
        return f"the price of the product is {self.__price}"

    
    #setter method-->set_price -->(new_price)-->should update only if new_price > 0
    def set_price(self, new_price):
        if(new_price > 0):
            self.__price = new_price
        return f"the new price is {self.__price}"

#test modifiying price using the setter
product_1 = Product("laptop", 75000, "electronics")
print(product_1.get_price())
print(product_1.set_price(100000))
print(product_1.get_price())
```

```python
#Task-3 inheritance(single-level)
#Create a subclass ElectronicProduct that inherits from Product
# Additional attribute:
# warranty_years  
class Product:

    def __init__(self, name, price, category):
        self.name = name 
        self.price = price 
        self.category = category

    #Override the get_info() method to include warranty info
    def get_info(self):
        print(f"name of the product {self.name} with the price {self.price} and the category is {self.category}")

class ElectronicProduc(Product):
    def __init__(self, name, price, category, warranty_years):
        super().__init__(name, price, category)
        self.warranty_years = warranty_years

    #Override the get_info() method to include warrenty info 
    def get_info(self):
        print(f"name of the product {self.name} with the price {self.price} and the category is {self.category} and we are also giving warranty of {self.warranty_years}years")

#create an object and demonstrate inheritance + overriding 
product_1 =  ElectronicProduc("laptop", 100000, "electronics")
product_1.get_info()

        
```

```python
##Task-4 Polymorphism
#create tow classes:

class Product:

    def __init__(self, name, price, category):
        self.name = name 
        self.price = price 
        self.category = category

#Laptop(Product)
#Both override:def get_info(self):
#Print details in their own style 
class Laptop(Product):

    def get_info(self):
        print(f"name:{self.name} | price:{self.price} | category:{self.category}")

#Mobile(Product)
#Both override:def get_info(self):
#Print details in their own style 
class Mobile(Product):

    def get_info(self):
        print(f"name:{self.name} | price:{self.price} | category:{self.category}")
    

product_1 = Laptop("Laptop", 100000, "Electronics")
product_2 = Mobile("Mobile", 50000,  "Electronics")

#write a loop that iterates over object of the laptop and Mobile and calls get_info() on each to show polymorphism
product = [product_1, product_2]
for i in product:
    i.get_info()
```

```python
#Task-5 Abstraction(using Abstract Base Class)
#Create an abstract class payment with abstract method 
#Process_payment(amount)
from abc import ABC, abstractmethod
class Payment(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass


#Then create two subclasses:
#CreditCardPayment
#UPIPayment
#Both override process_payment() with simple print statements
class CreditCardPayment(Payment):
   
    def process_payment(self, amount):
        print(f"payment done by creaditcard and the amount is: {amount}")

class UPIPayment(Payment):

    def process_payment(self, amount):
        print(f"payment done by upi and the amount is: {amount}")

#Test all classes
credit_1 = CreditCardPayment()
credit_1.process_payment(50000)
credit_2 = UPIPayment()
credit_2.process_payment(5000)
```

```python
#Task-7:Mini Project: Simple Inventory System(oop only)
#Create two classes:


class Product:

    def __init__(self, name, price, category):
        self.name = name 
        self.price = price 
        self.category = category

    def __str__(self):
            return f"{self.name} - {self.price} - {self.category}"

    def __add__(self, other):
        return Product(
            "Total",
            self.price + other.price,
            "Combined Price"
        )

    

#class: inventory
class Inventory:

    #Attributes:products → list to store product objects
    def __init__(self):
        self.products = []

    #Methods:
    #add_product(product)
    #remove_product(name)
    #get_total_value() → sums prices of all products
    #show_all_products() → prints info for each product
    def add_product(self, product):
        self.products.append(product)

    def remove_product(self, name):
        for i in self.products:
            if i.name == name:
                self.products.remove(i)
            break

    def get_total_value(self):
        total = 0
        for i in self.products:
            total += i.price
        return total 

    def show_all_products(self):
        for i in self.products:
            print("all_products:",i)


#Class: Store
class Store:

    #Attributes:
    # store_name
    # inventory → an Inventory object
    def __init__(self, store_name):
        self.store_name = store_name
        self.inventory = Inventory()

    #Methods:
    #add_new_product() → takes input & creates Product object
    #show_summary() → prints total items & value
    def add_new_product(self, name, price, category):
        product = Product(name, price, category)
        self.inventory.add_product(product)

    def show_summary(self):
        print("store_name:", self.store_name)
        print("total_items:", len(self.inventory.products))
        print("total_value:", self.inventory.get_total_value())

#Test the system by:
#Creating a Store object
store = Store("Praveen_Electronics")

#Adding 3 products
store.add_new_product("laptop", 75000, "Electronics")
store.add_new_product("keyboard", 2000, "Electronics")
store.add_new_product("mouse", 500, "Electronics")

#Showing summary
print("all_products")
store.inventory.show_all_products()
print("\nstore_summary")
store.show_summary()

#Using __add__ to combine prices of two products
product1 = store.inventory.products[0]
product2 = store.inventory.products[1]

print("\ncombined_products")
combined = product1 + product2
print(combined)




    
```
