```python
##task->1: safe Division Utility
#write a program that:

#1-> takes two inputs from the user:number and denominator
#2-> uses try-except to handle:
try:
    numerator = int(input("enter the numerator:"))
    denominator = int(input("enter the denominator:"))

    res = numerator/denominator
except ValueError as ex:
    print(ex)
except ZeroDivisionError as ax:
    print(ax)
    
 #3->if no error occurs print the result inside the else block
else:
    print("the result is:", res)
finally:
    print("operation complete")
```

```python
##Task->2: Bill Calculator with Error Handling
#Given a list of Product Prices:

prices = [120, 350, 'abc', 500, -200, 800]
result = 0

#1-> iterates through the list
for i in prices:
#2-> tries to add only valid(positive numerical) prices to the total
    try:
        if i < 0:
            raise ValueError("Negative price not allowed")
        else:
            result += i
        #4-> prints the running toatl
        print("running total:", result)

    #3-> Handles:
    #TypeError if value is not a number 
    #Custom exception using raise ValueError("Negative price not allowed")
    except TypeError as ex:
        print(ex)
    except ValueError as ax:
        print(ax)
         
               
```

```python
##Task->3 Custom Exception: Age Validator

#1->write a function check_age(age) that:
#raise a custom exception ValueError("Age must be between 1 and 120")
#if age is out of range 
def check_age(age):
    if age > 120 or age < 1:
        raise ValueError("Age must be between 1 and 120")
    else:
        print("age is in between 1 and 120")

#2-> in your main code:
#take age input from the user
age = int(input("enter the age:"))

#use try-except to catch and print the custom error message
try:
    #cheacking whether the age is  in between 1 and 120
    check_age(age)
except ValueError as ex:
    print(ex)
          
```

```python
#task->4 File Reader with Exception Handling
#1-> Ask the user for a filename
file_name = input("enter the file name to open:")

#2-> try to open and read the file 
try:
    with open(file_name, 'r') as f:
        content = f.read()

#3->Handle:
# FileNotFoundError
# PermissionError        
except FileNotFoundError as ex:
    print(ex)
except PermissionError as ax:
    print(ax)

#4->if successful, print first 3 lines of the file 
else:
    print("first three lines of the file:\n", content)

#5->use finally to print:
finally:
    print("File operation attempted")

```
