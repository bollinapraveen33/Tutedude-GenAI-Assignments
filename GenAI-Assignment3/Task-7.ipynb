```python
#Task-7 Mini Problem: Menu Using Functions
#Create the following three small functions:

#1.add_price(prices_list, price)-> adds price to the list
#2.get_average_price(prices_list)->return average price
#3.get_max_price(prices_lsit)->returns the maximum price

#i am using the statistics mean here to get the average 
#i am taking the list prices= [100, 250, 400, 1200, 50]
import statistics

prices = [100, 250, 400, 1200, 50]
def add_price(prices_list, price):
    prices_list.append(price)
    print("new element is appended in the list")

def get_average_price(prices_list):
   return statistics.mean(prices_list)

def get_max_price(prices_list):
    return max(prices_list)

#then creat a simple menu:
#1->add price
#2->show average price
#3->show highest price
#q->quit
#use only loops + function calls and conditional statements  

i=1
while i>0:
    print("option_1:appending or adding new value in the lsit")
    print("option_2:to get the average price")
    print("option_3:to get the max price")
    num = input("please choose the you option:")

    if num == '1':
        n= int(input("enter the new element you want append:"))
        add_price(prices, n)
    elif num == '2':
        print(get_average_price(prices))
    elif num == '3':
        print(get_max_price(prices))
    elif num == 'q':
        break
    else:
        print("you are enterd the worng option please enter a valued option")

```
