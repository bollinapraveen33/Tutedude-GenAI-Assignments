```python
#Task-3 Lambda function: GST Calculator

#Creat a lambda function gst that returns price after adding 18% GST
#Example:
#gst = lambda price: price + (0.18 * price)
gst = lambda price: price + (price * 0.18)
#print(gst(100))should return 118
print(gst(100))

#Extra(optional):Create another lambda to compute final price after GST + discount together
#I am taking the dicount as 5%
final_price = lambda price,discount: price + (price * 0.18)+ (price *discount/100)
#printing the final_price after adding the GST and discount
print(final_price(100,5))

```
