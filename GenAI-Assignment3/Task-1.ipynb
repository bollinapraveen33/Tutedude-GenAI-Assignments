```python
## Task-1 Basic Function: Price After Discount

#write a function apply_discount(price, discount_percent)
#1-> Returns the price after dicount
#2-> if dicount_percent is missing, apply a default discount of 5%
def apply_discount(price, discount_percent = 5):

    #extra(optional) add a condition inside the function to ensure discount never exceed 60%
    if discount_percent > 60:
        discount_percent = 60     

    discount_amount = price * discount_percent/100
    price_after_discount = price - discount_amount
    return price_after_discount


#3-> test the function with :
#4-> apply_dicount(1000, 10)
#5-> apply_dicount(500) uses default discout
print(apply_discount(1000, 75))
print(apply_discount(500))
```
